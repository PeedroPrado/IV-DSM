## Parte II – Distribuição de frequências

```
intervalos_memoria = [20,30,40,50,60,70,80,90,100]
```

```
import pandas as pd
def tabela_frequencia(dados, nome, intervalos):
  tabela = pd.cut(
  dados,
  bins=intervalos,
  right=False
  )
  freq = pd.Series(tabela).value_counts(sort=False)
  df = pd.DataFrame({
  "Intervalo": freq.index.astype(str),
  "Frequência": freq.values
  })
  df["Freq (%)"] = (
  df["Frequência"] / len(dados) * 100
  ).round(2)
  df["Acumulada"] = df["Frequência"].cumsum()
  print(nome)
  print(df)
  print("Total:", df["Frequência"].sum())
  print("-"*60)
  return df
```

```
tabela_A = tabela_frequencia(
servidor_A,
"Servidor A",
intervalos_memoria
)
tabela_B = tabela_frequencia(
servidor_B,
"Servidor B",
intervalos_memoria
)
tabela_C = tabela_frequencia(
servidor_C,
"Servidor C",
intervalos_memoria
)
tabela_D = tabela_frequencia(
servidor_D,
"Servidor D",
intervalos_memoria
)
```

## Parte III – Média, variância e desvio padrão

```
import statistics as st
servidores = {
"A": servidor_A,
"B": servidor_B,
"C": servidor_C,
"D": servidor_D
}
for nome, dados in servidores.items():
  media = st.mean(dados)
  variancia = st.pvariance(dados)
  desvio = st.pstdev(dados)
  print("Servidor", nome)
  print("Média:", round(media,2))
  print("Variância:", round(variancia,2))
  print("Desvio padrão:", round(desvio,2))
  print("-"*60)
```
