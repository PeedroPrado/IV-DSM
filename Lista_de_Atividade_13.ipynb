Parte 2 – Distribuição de frequências# Nova secção

```
import pandas as pd
def tabela_frequencia(dados, nome):
    intervalos = [20, 30, 40, 50, 60, 70, 80, 90]
    tabela = pd.cut(dados, bins=intervalos, right=False)
    freq = pd.Series(tabela).value_counts(sort=False)
    df = pd.DataFrame({
        "Intervalo": freq.index.astype(str),
        "Frequência": freq.values
    })
    df["Frequência relativa (%)"] = (df["Frequência"] / len(dados) * 100).round(2)
    df["Frequência acumulada"] = df["Frequência"].cumsum()
    print(nome)
    print(df)
    print("Total de registros:", df["Frequência"].sum())
    print("-"*50)
    return df
```

```
tabela_A = tabela_frequencia(sistema_A, "Sistema A")
tabela_B = tabela_frequencia(sistema_B, "Sistema B")
tabela_C = tabela_frequencia(sistema_C, "Sistema C")
tabela_D = tabela_frequencia(sistema_D, "Sistema D")
```

# Parte 3 – Média, variância e desvio padrão

```
import statistics as st
sistemas = {
"A": sistema_A,
"B": sistema_B,
"C": sistema_C,
"D": sistema_D
}
for nome, dados in sistemas.items():
  print("Sistema", nome)
  print("Média:", round(st.mean(dados), 2))
  print("Variância:", round(st.pvariance(dados), 2))
  print("Desvio padrão:", round(st.pstdev(dados), 2))
  print("-"*50)
```
